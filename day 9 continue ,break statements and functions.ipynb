### Basic calculator using functions

```python
def add(a,b):
    x=a+b
    return x
def sub(a,b):
    y=a-b
    return y
def mul(a,b):
    z=a*b
    return z
def div(a,b):
    if b==0 :
        return "zero division not possible"
    u=a/b
    return u

if __name__ == "__main__" :
    print("welcome to small level calculator") 

    
x=add(40,20)
print(x)
y=sub(50,20)
print(y)
z=mul(20,20)
print(z)
u=div(40,20)
print(u)
```

```python
def addition(a,b):
    return a+b
def subtraction(a,b):   
    return a-b
def multiple(a,b):   
    return a*b
def division(a,b):
    if b==0:
        return "Zero division not possible"
    return a/b
if __name__ == "__main__":
    print("Welcome to small level calculator")
    n1 = int(input("Enter number 1:"))
    n2 = int(input("Enter number 2:"))
    print("1. addition /n 2. subtraction /n 3. multiple /n 4. division")
    choice = int(input("Enter your choice:"))
    if choice == 1:
        print("addition of two numbers s:", addition(n1,n2))
    elif choice == 2:
        print("subtraction of two numbers s:", subtraction(n1,n2))
    elif choice == 3:
        print("multiple of two numbers s:", multiple(n1,n2))
    elif choice == 4:
        print("division of two numbers s:", division(n1,n2))
    else :
        print("please select correct operation")
```
