3.keyword Arguments
4 variable lengths positional of arguments
5 variable length of keyword arguments

```python
# positional arguments 
#subtraction of two numbers using functions
def sub (n1,n2):
    res = n1 - n2 
    return res 
a = int(input())  
b = int(input()) # a,b = 10,20
print(sub(a, b))
print(sub(b, a))
```

```python
# keyword arguments
# subtraction of two numbers using functions
def sub(n1,n2):
    res = n1 - n2 
    return res 
a = int(input())  
b = int(input())
print(sub(a, b))
print(sub(b, a))
```

```python
#2. sum of 4 numbers usig functions
def Adding_four_numbers(z1,z2,z3,z4):
    res = z1 + z2 + z3 + z4
    return res
a = int(input("Enter a number 1:"))
b = int(input("Enter a number 2:"))
c = int(input("Enter a number 3:"))
d = int(input("Enter a number 4:"))
res = Adding_four_numbers(a,b,c,d)
print(res)
```

### List properties

```python
## list properties
# list elements access by using index positions
lst = [10,2,4,4.534, 'hi','hello']
print(lst[2])
print(lst[-3])
print(lst[2:])
```

```python
# list mutable
lst = [1,2,3,40,'hi','hello']
lst[0] = 100
print(lst)
```

```python
# string are immutable
str="asdfgh"
str[0]= "R"
print(str)
```

### nested list element access

```python
# nested list element access
lst =[1,2,3,4,[10,30,50, ['a','b','c',]],29.1,3]
print(lst[4])
print(lst[4][3])
print(lst[4][3][2])
```

# read list of numbers if numer is even add 0.5 and if the number is odd add one

```python
#read list of numbers if numer is even add 0.5 and if the number is odd add one
##lst=[1,2,3,4,5]
lst=list(map(int,input().split()))
for i in range(0,len(lst)):
    if lst[i]%2 :
        lst[i]=lst[i]+1
    else:
         lst[i]=lst[i]+0.5
print(lst)
```
