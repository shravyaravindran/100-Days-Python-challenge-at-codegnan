```python
# hallow right angle triangle 
n=6
for i in range(n):
    for j in range(n):
        if i == j or j==0 or i==n-1 :
            print("*", end = " ")
        else:
            print(" ", end = " ")
    print()  
```

```python
# Reverse hallow right angle triangle 
n=6
for i in range(n):
    for j in range(n):
        if i == j or i==0 or j==n-1 :
            print("*", end = " ")
        else:
            print(" ", end = " ")
    print()  
```
